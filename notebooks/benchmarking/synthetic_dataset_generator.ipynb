```python
# Run this cell to create a "True Leak" benchmark file inside your notebook
import pandas as pd
import numpy as np
import os

# Create 90 days of hourly timestamps
dates = pd.date_range(start="2026-01-01", periods=90*24, freq="h")

# Generate standard residential behavior (low at night, spikes in morning/evening)
base_use = []
for d in dates:
    if 1 <= d.hour <= 4:
        base_use.append(np.random.uniform(0.000, 0.005)) # Background noise near zero
    elif 7 <= d.hour <= 9 or 18 <= d.hour <= 21:
        base_use.append(np.random.uniform(0.150, 0.300)) # Peak shower/cooking hours
    else:
        base_use.append(np.random.uniform(0.020, 0.060)) # Mid-day usage

test_df = pd.DataFrame({'Hourly': dates, 'Consumption m3': base_use})

# INJECT A CRITICAL LEAK: Force a slow leak of 40 liters/hour (0.040 m3) into the last week
test_df.loc[test_df['Hourly'] >= "2026-03-24", 'Consumption m3'] += 0.040

# Save it to your folder as a zero-header benchmark file
os.makedirs("datasets", exist_ok=True)
test_df.to_csv("datasets/synthetic_residential_leak.csv", index=False)
print("Created standard benchmark file with an injected leak at 'datasets/synthetic_residential_leak.csv'!")
```
